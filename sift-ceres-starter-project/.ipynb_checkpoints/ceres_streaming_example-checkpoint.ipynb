```python
# Standard library imports
import os
from datetime import datetime, timezone
from zoneinfo import ZoneInfo
from random import random
from time import sleep, time
from typing import Iterator, Tuple

# Third-party imports
import pandas as pd
import xarray as xr

# Sift-related imports
from sift_py.grpc.transport import SiftChannelConfig, use_sift_channel
from sift_py.ingestion.channel import (
    ChannelConfig, 
    ChannelDataType, 
    double_value, 
    float_value
)
from sift_py.ingestion.config.telemetry import TelemetryConfig
from sift_py.ingestion.flow import FlowConfig
from sift_py.ingestion.service import IngestionService
from sift_py.calculated_channels.service import CalculatedChannelService
from sift_py.calculated_channels.config import CalculatedChannelConfig
from sift_py.grpc.transport import SiftChannelConfig, use_sift_channel
```

```python
# STEP 1: Import CERES NetCDF file and process it into a DataFrame that can represent a stream.

file_path = os.path.expanduser('~/dev/jonno-hacks/sift-ceres-starter-project/CERES_SYN1deg-1H_Terra-Aqua-MODIS_Ed4.1_Subset_20000301-20240731.nc')

try:
    ds = xr.open_dataset(file_path, engine='netcdf4')
    print(f"Successfully loaded dataset from: {file_path}")
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
    print("Please ensure the file path is correct and the file exists.")
    ds = None # Set ds to None to avoid errors if the file wasn't loaded
except Exception as e:
    print(f"An error occurred while opening the dataset: {e}")
    ds = None

if ds is not None:
    # --- Step 5: Inspect the dataset structure ---
    print("\n--- Dataset Information (ds) ---")
    print(ds)

    print("\n--- Dataset Variables (ds.data_vars) ---")
    print(ds.data_vars)

    # --- Define the specific variable names from your NetCDF file ---
    # These are both 'clear-sky' longwave up fluxes
    toa_clear_sky_lw_flux_name = 'adj_atmos_lw_up_clr_toa_1h'
    surface_clear_sky_lw_flux_name = 'adj_atmos_lw_up_clr_surface_1h'

    if toa_clear_sky_lw_flux_name in ds and surface_clear_sky_lw_flux_name in ds:
        toa_clear_sky_flux = ds[toa_clear_sky_lw_flux_name]
        surface_clear_sky_flux = ds[surface_clear_sky_lw_flux_name]

        la_combined_data = xr.Dataset({
            'toa_clear_sky_lw': toa_clear_sky_flux,
            'surface_clear_sky_lw': surface_clear_sky_flux
        })
        df_la = la_combined_data.to_dataframe().reset_index()

        # Ensure the 'time' column is a timezone-aware UTC datetime object
        la_tz = ZoneInfo("America/Los_Angeles")
        
        # Function to convert naive datetime to UTC
        def convert_to_utc(dt_obj):
            if dt_obj.tzinfo is None: # If datetime is naive
                # Localize to LA timezone, then convert to UTC
                return dt_obj.replace(tzinfo=la_tz).astimezone(timezone.utc)
            else: # If datetime is already timezone-aware (e.g., already UTC)
                return dt_obj.astimezone(timezone.utc)

        df_la['time_utc'] = df_la['time'].apply(convert_to_utc)
        
        df_la.drop(columns=['time'], inplace=True)
        df_la.rename(columns={'time_utc': 'timestamp'}, inplace=True)


        print("\n--- Final Pandas DataFrame for LA (first 5 rows) ---")
        print(df_la.head())
        print(f"\nTotal rows in final LA DataFrame: {len(df_la)}")

    else:
        print(f"\nCould not find expected variables in the dataset:")
        print(f"  - '{toa_clear_sky_lw_flux_name}'")
        print(f"  - '{surface_clear_sky_lw_flux_name}'")
        print("Please check the `ds.data_vars` output above to confirm the exact variable names.")
```

```python
# STEP 2: Ingest the stream as a Run in Sift

# The name of our asset
ASSET_NAME = "CERES_longwave_up_flux"
 
# An arbitrary key used to identify our config
CONFIG_KEY = "CERES_longwave_up_flux-config-v5"
 
# Define the schema of your telemetry
config = TelemetryConfig(
    asset_name=ASSET_NAME,
    ingestion_client_key=CONFIG_KEY,
    flows=[
        FlowConfig(
            name="CERES_longwave_up_flux",
            channels=[
                ChannelConfig(
                    name="surface_clear_sky_flux",
                    data_type=ChannelDataType.FLOAT,
                    unit="W/m2",
                ),
                ChannelConfig(
                    name="toa_clear_sky_flux",
                    data_type=ChannelDataType.FLOAT,
                    unit="W/m2",
                ),
            ],
        ),
    ],
)

data_source = df_la

local_credentials: SiftChannelConfig = {
    "apikey": "1CTVOwfktFMNlWvctYte3jKoA61ub0S0AlDKfaQg",
    "uri": "http://localhost:50051",
    "use_ssl": False,
}
dev_credentials = {
    "apikey": "PlhxwmIsTvou5w6o9MHk5ttXECAWa7MZjfreVJhx",
    "uri": "https://grpc-api.development.siftstack.com",
}
 
 
with use_sift_channel(dev_credentials) as grpc_channel:
    # Create an ingestion service
    ingestion_service = IngestionService(grpc_channel, config)
 
    # Create a run to use during this period of ingestion
    run_name = f"[{ASSET_NAME}].{time():.0f}"
    ingestion_service.attach_run(grpc_channel, run_name)
    print('channel created')

    # Stream sensor data to Sift
    for idx, row in data_source.iterrows():
    
        channel_values = [
            float_value(row['surface_clear_sky_lw']),
            float_value(row['toa_clear_sky_lw']),
        ]

        ingestion_service.ingest_flows({
            "flow_name": "CERES_longwave_up_flux",
            "timestamp": row['timestamp'],
            "channel_values": channel_values
        })
```

```python
# STEP 3: use gRPC API to create a new calculated channel via API:
channel_references = [
     {
        "channel_reference": "$x",
        "channel_identifier": "top_of_atmosphere_lw_up_flux",
    },
    {
        "channel_reference": "$y",
        "channel_identifier": "surface_lw_up_flux",
    },
]

# arbitrary client id to identify this calculated channel
client_key = "toa_minus_surface_lw_flux_v4"

config = CalculatedChannelConfig(
    name="difference-toa-surface-lw-flux-v4",
    description="Difference between TOA and surface measurements for infrared LW radiation as a proxy for heat retention by Earth's atmosphere.",
    expression="$y - $x",
    channel_references=channel_references,
    units="W/m2",
    client_key=client_key,
    all_assets=True,
)

 
with use_sift_channel(dev_credentials) as grpc_channel:
    channel_service = CalculatedChannelService(grpc_channel)

    channel_service.create_calculated_channel(config)
  
```
